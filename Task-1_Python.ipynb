## Q35

**Prediction**

[[1, 99], [3, 4]]

**Explanation** A shallow copy duplicates only the outer list inner lists are shared, so modifying an inner list affects a.

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
````

## Q36

**Prediction**

[[1, 2], [3, 4]]

**Explanation:** A deep copy recursively copies nested lists, so changing b's inner list does not change a.

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
````

## Q38
